```python
import thermomatch as match
match.ThermoImpexGenerator.create_local_database_if_not_existent = True
```

```python
# Set path to `schemas` and `lua` not from the configuration file
match.ThermoImpexGenerator.setResourcesDirectory("../../../../Resources")

# Connect to Arango database
generator = match.ThermoImpexGenerator("http://localhost:8529", "root", "", "ORD_heracles", True, match.ThermoImpexGenerator.File, False)
```

```python
# List of implemented thermodynamic formats
#print( generator.allFormats() )

# Execute import from PSI table format data
description = [
    match.ImpexRowDescription("refs", "VertexDataSource", "records_and_links", "FormatStructDataFile", "scripts/references.FormatStructDataFile.json"), # Datasource records 
    match.ImpexRowDescription("elements", "VertexElement", "records", "FormatStructDataFile", "scripts/Element.heracles.FormatStructDataFile.json"), # elements records
    match.ImpexRowDescription("substances2", "VertexSubstance", "records", "FormatStructDataFile", "scripts/Substance.heracles.FormatStructDataFile.json"), # substances records
    match.ImpexRowDescription("reactions", "VertexSubstance", "records", "FormatStructDataFile", "scripts/ReactionSubstance.heracles.FormatStructDataFile.json"), # reaction dependent substances records
    match.ImpexRowDescription("reactions", "VertexReaction", "group", "FormatStructDataFile", "scripts/Reaction.heracles.FormatStructDataFile.json"), # reaction records
    match.ImpexRowDescription("reactions", "EdgeTakes", "group", "FormatStructDataFile", "scripts/ReactionEdge.FormatStructDataFile.json"), # links between reactants and reaction records
    match.ImpexRowDescription("phasespure", "VertexPhase", "records_and_links", "FormatStructDataFile", "scripts/Phase.heracles.FormatStructDataFile.json"), # phase records
    match.ImpexRowDescription("ThermoDataSet", "VertexThermoDataSet", "records_and_links", "FormatTableFile", "scripts/thermodataset.FormatTableFile.json") # thermodataset record
]
```

```python
#print( description )
generator.runImport("ReacDCImport", description, "data/heracles.refs.json", "");
```
