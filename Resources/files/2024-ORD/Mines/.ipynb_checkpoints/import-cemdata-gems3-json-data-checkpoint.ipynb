```python
# List of implemented thermodynamic formats
#print( generator.allFormats() )

# Execute import from PSI table format data
description = [
    match.ImpexRowDescription("refs", "VertexDataSource", "records_and_links", "FormatStructDataFile", "scripts/references.FormatStructDataFile.json"), # Datasource records 
    match.ImpexRowDescription("elements", "VertexElement", "records", "FormatStructDataFile", "scripts/Element.mines23.FormatStructDataFile.json"), # elements records
    match.ImpexRowDescription("aqueous", "VertexSubstance", "records", "FormatStructDataFile", "scripts/Substance.mines23.FormatStructDataFile.json"), # aqueous substances records
    match.ImpexRowDescription("solids", "VertexSubstance", "records", "FormatStructDataFile", "scripts/Substance.mines23.FormatStructDataFile.json"), # solid substances records
    match.ImpexRowDescription("fluids", "VertexSubstance", "records", "FormatStructDataFile", "scripts/Substance.mines23.FormatStructDataFile.json"), # fluid substances records
    match.ImpexRowDescription("reactions", "VertexSubstance", "records", "FormatStructDataFile", "scripts/ReactionSubstance.mines23.FormatStructDataFile.json"), # reaction dependent substances records
    match.ImpexRowDescription("reactions", "VertexReaction", "group", "FormatStructDataFile", "scripts/Reaction.mines23.FormatStructDataFile.json"), # reaction records
    match.ImpexRowDescription("reactions", "EdgeTakes", "group", "FormatStructDataFile", "scripts/ReactionEdge.mines23.FormatStructDataFile.json"), # links between reactants and reaction records
    match.ImpexRowDescription("phasespure", "VertexPhase", "records_and_links", "FormatStructDataFile", "scripts/Phase.mines23.FormatStructDataFile.json"), # phase records
    match.ImpexRowDescription("ThermoDataSet", "VertexThermoDataSet", "records_and_links", "FormatTableFile", "scripts/thermodataset.FormatTableFile.json") # thermodataset record
]
```

```python
#print( description )
generator.runImport("ReacDCImport", description, "data/mines23.refs.json", "mines23");
```
