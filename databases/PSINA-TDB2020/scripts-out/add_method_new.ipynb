```python
import json

# Define the default TPMethods block
default_tp_method = [
    {
        "method": {
            "43": "standard_entropy_cp_integration "
        }
    }
]

# Load the original JSON file
with open("master.VertexSubstance.json", "r") as f:
    data = json.load(f)

# Loop through substances and add TPMethods if missing or empty
for substance in data:#.get("substances", []):
    props = substance.get("properties")
    if not props.get("TPMethods") and "_rdc_" not in props.get("name", ""):
        props["TPMethods"] = default_tp_method

# Save to a new JSON file
with open("masterx.VertexSubstance.json", "w") as f:
    json.dump(data, f, indent=2)
```
